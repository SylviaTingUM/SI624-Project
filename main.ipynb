## Project Overview

Medicare represents a substantial component of U.S. healthcare expenditures. According to the Centers for Medicare & Medicaid Services (CMS), Medicare spending reached approximately $1.03 trillion in 2023, accounting for 21% of national health expenditures and increasing by 8.1% from the previous year (CMS, 2024). Understanding geographic differences in Medicare spending is therefore important for healthcare policymakers and researchers seeking to examine spending patterns and their potential socioeconomic associations.

This project investigates how county-level socioeconomic conditions relate to standardized Medicare spending per capita across U.S. counties. We combine the CMS Medicare Geographic Variation dataset with the U.S. Census Bureau's 2023 American Community Survey (ACS) 5-Year Estimates. The CMS data provide information on Original Medicare spending and beneficiary characteristics, while the ACS data provide socioeconomic indicators, including median household income, poverty rate, and unemployment rate.

The merged dataset contains 3,143 county-level geographic observations, with 3,140 complete observations available for the primary exploratory analysis. We use descriptive statistics, correlation analysis, visualizations, and multiple linear regression to examine the relationships between socioeconomic conditions and Medicare spending while considering beneficiary characteristics and Medicare Advantage participation.

Preliminary exploratory findings indicate that counties with higher poverty rates tend to have higher standardized Medicare spending per capita, while median household income shows a weaker negative association with spending. Because the analysis uses observational county-level data, the findings describe associations rather than causal effects.

### Data Overview

#### Data Sources

This project combines two publicly available datasets to examine the relationship between socioeconomic conditions and Medicare spending across U.S. counties.

**Dataset 1: Medicare Geographic Variation**

The first dataset is the Medicare Geographic Variation public-use file from the Centers for Medicare & Medicaid Services (CMS). It contains information on Medicare spending, health-care utilization, and beneficiary characteristics at the national, state, and county levels.

We use county-level observations for 2023, focusing on beneficiaries enrolled in Original Medicare (fee-for-service). The primary outcome variable is standardized Medicare payment per capita (`TOT_MDCR_STDZD_PYMT_PC`), which allows comparisons of Medicare spending across geographic areas.

Additional variables include average beneficiary age (`BENE_AVG_AGE`), dual-eligible percentage (`BENE_DUAL_PCT`), and Medicare Advantage participation rate (`MA_PRTCPTN_RATE`).

**Dataset 2: American Community Survey (ACS)**

The second dataset is the 2023 American Community Survey (ACS) 5-Year Estimates from the U.S. Census Bureau.

The ACS provides demographic and socioeconomic information for U.S. counties. We use median household income, poverty rate, and unemployment rate as the primary socioeconomic indicators.

The 2023 ACS 5-Year Estimates combine survey data collected from 2019 through 2023, while the CMS observations represent 2023. Therefore, the two datasets are aligned by reference year but do not represent identical data collection periods.

The datasets are combined using five-digit county FIPS codes.

#### Unit of Observation and Population

The unit of observation is a U.S. county or county-equivalent geographic area. Each row in the merged dataset represents one geographic unit with Medicare spending information from CMS and socioeconomic characteristics from ACS.

The Medicare spending measures describe beneficiaries enrolled in Original Medicare (fee-for-service), rather than the entire U.S. population or all Medicare beneficiaries. The ACS socioeconomic measures describe the broader county population.

#### Data Access and Usage

**Dataset 2: American Community Survey (ACS)**

The 2023 ACS 5-Year Estimates were obtained from the U.S. Census Bureau through the Census API. The data represent the 2019–2023 survey period.

Official source: [American Community Survey 5-Year Data (2023)](https://www.census.gov/data/developers/data-sets/acs-5year/2023.html)

**ACS Data Retrieval and Variable Construction**

The 2023 ACS 5-Year Estimates were retrieved through the U.S. Census Bureau Data API. The original variables include median household income (`B19013_001E`), the population for whom poverty status is determined (`B17001_001E`), the population below the poverty level (`B17001_002E`), the civilian labor force (`B23025_003E`), and the unemployed civilian labor force (`B23025_005E`).

Poverty rate was calculated as the population below the poverty level divided by the population for whom poverty status is determined, multiplied by 100.

Unemployment rate was calculated as the unemployed civilian labor force divided by the total civilian labor force, multiplied by 100.

County FIPS codes were constructed by combining the two-digit state code and three-digit county code. The resulting dataset was saved as a preprocessed CSV file containing county FIPS, county name, median household income, poverty rate, and unemployment rate.

The preprocessed ACS CSV is included in the submitted `data/` folder so that the main notebook can be reproduced without making additional API requests.

**Data Usage and Terms of Use**

The CMS Medicare Geographic Variation dataset is publicly available and classified as a Public Use File (Free).

The terms permit accessing and analyzing Census data but prohibit attempts to identify individual respondents or other protected entities.

Both datasets contain aggregated geographic information rather than individual-level records. This project uses the data for academic research and does not attempt to identify individuals.

The ACS socioeconomic indicators were derived from Census estimates, and the resulting variables are identified as calculated measures rather than original Census variables.

### Data Preparation and Cleaning

#### Filtering the CMS Dataset

The CMS dataset contains observations from multiple years and geographic levels. To match the research question, we restrict the data to:

- `YEAR == 2023`
- `BENE_GEO_LVL == "County"`
- `BENE_AGE_LVL == "All"`

The age-level filter retains observations representing all eligible beneficiaries rather than separate age subgroups.

After filtering, the CMS dataset contains 3,197 county-level geographic records for 2023.

This step establishes the geographic and temporal scope of the analysis before integrating socioeconomic information from ACS.

```python
cms_2023 = cms[
    (cms["YEAR"] == 2023) &
    (cms["BENE_GEO_LVL"] == "County") &
    (cms["BENE_AGE_LVL"] == "All")
].copy()

print(cms_2023.shape)
```

#### Standardizing County FIPS Codes

The CMS and ACS datasets use county FIPS codes to identify geographic units.

To ensure consistent matching, CMS geographic codes are converted to five-character strings with leading zeros preserved. The ACS FIPS identifiers are standardized to the same format.

Preserving leading zeros is necessary because some county FIPS codes begin with zero. Treating these identifiers as ordinary integers could cause incorrect geographic matching.

The standardized FIPS variable is used as the common key for merging the datasets.

```python
cms_2023["fips"] = (
    cms_2023["BENE_GEO_CD"]
    .astype("Int64")
    .astype(str)
    .str.zfill(5)
)
print(
    cms_2023[
        ["BENE_GEO_CD", "fips", "BENE_GEO_DESC"]
    ].head(10)
)
```

```python
acs["fips"] = (
    acs["fips"]
    .astype(str)
    .str.zfill(5)
)
```

#### Assessment of Extreme Values

We assess potential extreme values in the four primary analysis variables using the interquartile range (IQR) method.

For each variable, observations below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are flagged as potential outliers.

This method provides a consistent way to identify unusually high or low values without assuming a normal distribution.

However, statistical outliers are not necessarily data errors. Extreme values may reflect genuine geographic variation in Medicare spending and socioeconomic conditions.

Therefore, potential outliers are identified and documented but are not automatically removed from the analysis.

```python
# Identify potential extreme values using the 1.5 × IQR rule

outlier_variables = [
    "medicare_spending",
    "median_household_income",
    "poverty_rate",
    "unemployment_rate"
]

outlier_results = []

for variable in outlier_variables:
    q1 = analysis_df[variable].quantile(0.25)
    q3 = analysis_df[variable].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_count = (
        (analysis_df[variable] < lower_bound) |
        (analysis_df[variable] > upper_bound)
    ).sum()

    outlier_results.append({
        "Variable": variable,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Potential Outliers": outlier_count
    })

outlier_summary = pd.DataFrame(outlier_results)

outlier_summary
```

```python
validity_check = pd.DataFrame({
    "Variable": [
        "medicare_spending",
        "median_household_income",
        "poverty_rate",
        "unemployment_rate"
    ],
    "Invalid Values": [
        (analysis_df["medicare_spending"] < 0).sum(),
        (analysis_df["median_household_income"] <= 0).sum(),
        (
            (analysis_df["poverty_rate"] < 0) |
            (analysis_df["poverty_rate"] > 100)
        ).sum(),
        (
            (analysis_df["unemployment_rate"] < 0) |
            (analysis_df["unemployment_rate"] > 100)
        ).sum()
    ]
})

validity_check
```

#### Correlation Analysis

We calculate Pearson correlation coefficients to examine the direction and strength of linear associations between Medicare spending and the three socioeconomic indicators.

Pearson correlation coefficients range from -1 to 1.

Positive values indicate that two variables tend to increase together, while negative values indicate that one variable tends to decrease as the other increases.

Values closer to zero indicate weaker linear relationships.

This analysis provides an initial assessment of socioeconomic associations before accounting for additional beneficiary characteristics in the regression models.

Because correlation does not establish causation, the results are interpreted as descriptive associations.

**Assumptions and Limitations**

Pearson correlation measures linear associations between continuous variables. The scatterplots presented below help assess the form of these relationships.

Each county is treated as one observation. However, nearby counties may share economic and healthcare characteristics, so geographic dependence cannot be ruled out.

The correlations are descriptive and do not adjust for potential confounding factors. Therefore, they should not be interpreted as causal relationships.

```python
correlation = analysis_df[
    [
        "medicare_spending",
        "median_household_income",
        "poverty_rate",
        "unemployment_rate"
    ]
].corr()

correlation
```

**Confidence Intervals for Pearson Correlations**

To quantify uncertainty in the estimated correlations, we calculate approximate 95% confidence intervals using Fisher's z transformation.

The intervals are based on the number of complete county-level observations for each variable pair.

These intervals assume approximately independent observations and do not account for potential spatial dependence between counties. Therefore, they should be interpreted as approximate measures of uncertainty.

```python
correlation_variables = [
    "median_household_income",
    "poverty_rate",
    "unemployment_rate"
]

confidence_results = []

for variable in correlation_variables:
    paired_data = analysis_df[
        [variable, "medicare_spending"]
    ].dropna()
    n = len(paired_data)
    r = paired_data[variable].corr(
        paired_data["medicare_spending"]
    )
    fisher_z = np.arctanh(r)
    standard_error = 1 / np.sqrt(n - 3)
    z_lower = fisher_z - 1.96 * standard_error
    z_upper = fisher_z + 1.96 * standard_error
    confidence_results.append({
        "Variable": variable,
        "N": n,
        "Pearson r": r,
        "95% CI Lower": np.tanh(z_lower),
        "95% CI Upper": np.tanh(z_upper)
    })

correlation_ci = pd.DataFrame(confidence_results)
correlation_ci.round(3)
```

**Interpretation of Correlation Results**

The correlation analysis shows different relationships between Medicare spending and the three socioeconomic indicators.

Median household income has a weak negative correlation with Medicare spending (r = -0.199), suggesting that counties with higher household income tend to have somewhat lower standardized Medicare spending.

Poverty rate has a positive correlation with Medicare spending (r = 0.258), indicating that counties with higher poverty rates tend to have higher spending.

Unemployment rate has a much weaker positive correlation with Medicare spending (r = 0.098).

Among the three socioeconomic indicators, poverty rate has the strongest linear correlation with Medicare spending in absolute terms.

The socioeconomic variables are also correlated with one another. In particular, median household income and poverty rate have a relatively strong negative correlation (r = -0.729).

These findings suggest that socioeconomic conditions are associated with Medicare spending, but the relationships are modest and may be influenced by other county-level characteristics.

The regression analysis will examine these associations while accounting for additional variables.

The approximate 95% confidence intervals are [-0.232, -0.165] for median household income, [0.225, 0.291] for poverty rate, and [0.063, 0.132] for unemployment rate.

All three intervals exclude zero, suggesting that the observed linear associations are distinguishable from zero under the assumptions of the Fisher z method. However, the magnitude of these correlations remains relatively small, particularly for unemployment rate.

These intervals assume approximately independent county observations and do not account for potential spatial dependence or other sources of bias. The findings remain descriptive associations rather than evidence of causation.

#### Model 1

```python
import statsmodels.api as sm

regression_df = analysis_complete[
    [
        "medicare_spending",
        "median_household_income",
        "poverty_rate",
        "unemployment_rate"
    ]
].copy()

# Express household income in units of $10,000
regression_df["income_10k"] = (
    regression_df["median_household_income"] / 10000
)

# Independent variables
X = regression_df[
    [
        "income_10k",
        "poverty_rate",
        "unemployment_rate"
    ]
]

# Dependent variable
y = regression_df["medicare_spending"]

# Add intercept
X = sm.add_constant(X)

# Fit multiple linear regression with robust standard errors
model_1 = sm.OLS(y, X).fit(cov_type="HC3")

print(model_1.summary())
```

Model 1 explains about 6.9% of the variation in Medicare spending. Poverty rate is positively associated with spending and is statistically significant, while median household income is not statistically significant. Unemployment has a small negative association with spending after controlling for the other socioeconomic variables.

#### Model 2

```python
regression_df_2 = analysis_complete[
    [
        "medicare_spending",
        "median_household_income",
        "poverty_rate",
        "unemployment_rate",
        "BENE_AVG_AGE",
        "BENE_DUAL_PCT"
    ]
].copy()

# Express household income in units of $10,000
regression_df_2["income_10k"] = (
    regression_df_2["median_household_income"] / 10000
)

X2 = regression_df_2[
    [
        "income_10k",
        "poverty_rate",
        "unemployment_rate",
        "BENE_AVG_AGE",
        "BENE_DUAL_PCT"
    ]
]

y2 = regression_df_2["medicare_spending"]

X2 = sm.add_constant(X2)

model_2 = sm.OLS(y2, X2).fit(cov_type="HC3")

print(model_2.summary())
```

After adding average beneficiary age and the percentage of dual-eligible beneficiaries, the model explains about 10.2% of the variation in Medicare spending. Poverty rate, average beneficiary age, and dual-eligible percentage are positively associated with spending and statistically significant. In contrast, income and unemployment are not statistically significant in the adjusted model. The increase in R² from 6.9% to 10.2% suggests that beneficiary characteristics provide additional explanatory power beyond the socioeconomic variables.
